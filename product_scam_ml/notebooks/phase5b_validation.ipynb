```python
# ============================================================
# PHASE 5B — REAL RAW AMAZON → FASTAPI TEST
# ============================================================

import requests
import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Take one REAL raw Amazon product
# ------------------------------------------------------------

raw_product = raw_df.iloc[0].to_dict()


# ------------------------------------------------------------
# Convert pandas NaN → None
#
# This is necessary because JSON cannot represent NaN
# reliably in a standard API request.
# ------------------------------------------------------------

def clean_for_json(value):

    if isinstance(value, dict):

        return {
            k: clean_for_json(v)
            for k, v in value.items()
        }

    if isinstance(value, list):

        return [
            clean_for_json(v)
            for v in value
        ]

    if isinstance(value, np.ndarray):

        return [
            clean_for_json(v)
            for v in value.tolist()
        ]

    if pd.isna(value):

        return None

    return value


raw_product = clean_for_json(
    raw_product
)


# ------------------------------------------------------------
# Send REAL Amazon product to FastAPI
# ------------------------------------------------------------

API_URL = "http://127.0.0.1:8000/predict"

response = requests.post(
    API_URL,
    json=raw_product,
    timeout=30
)

print("=" * 80)
print("REAL AMAZON PRODUCT → FASTAPI")
print("=" * 80)

print(f"HTTP Status: {response.status_code}")

response.raise_for_status()

api_result = response.json()


# ------------------------------------------------------------
# Display result
# ------------------------------------------------------------

print("\nPRODUCT")
print("-" * 80)

print(
    "Parent ASIN:",
    raw_product.get("parent_asin")
)

print(
    "Title:",
    raw_product.get("title")
)

print("\nRISK RESULT")
print("-" * 80)

print(
    json.dumps(
        api_result,
        indent=4
    )
)
```

```python
# ============================================================
# PHASE 5B — MULTI-PRODUCT FASTAPI PARITY VALIDATION
# ============================================================

import requests
import numpy as np
import pandas as pd
from IPython.display import display

API_URL = "http://127.0.0.1:8000/predict"

# ------------------------------------------------------------
# Number of real products to test
# ------------------------------------------------------------

N_TEST = 50

test_df = raw_df.head(N_TEST).copy()


# ------------------------------------------------------------
# Convert pandas / NumPy values into JSON-safe values
# ------------------------------------------------------------

def clean_for_json(value):

    if isinstance(value, dict):

        return {
            k: clean_for_json(v)
            for k, v in value.items()
        }

    if isinstance(value, list):

        return [
            clean_for_json(v)
            for v in value
        ]

    if isinstance(value, tuple):

        return [
            clean_for_json(v)
            for v in value
        ]

    if isinstance(value, np.ndarray):

        return [
            clean_for_json(v)
            for v in value.tolist()
        ]

    if isinstance(value, (np.integer,)):

        return int(value)

    if isinstance(value, (np.floating,)):

        if np.isnan(value):
            return None

        return float(value)

    if pd.isna(value):

        return None

    return value


# ------------------------------------------------------------
# Run DIRECT Python engine + FastAPI for every product
# ------------------------------------------------------------

parity_results = []

for index, (_, row) in enumerate(test_df.iterrows()):

    raw_product = clean_for_json(
        row.to_dict()
    )

    # --------------------------------------------------------
    # DIRECT PYTHON ENGINE
    # --------------------------------------------------------

    direct_result = predict_product_risk(
        raw_product
    )

    # --------------------------------------------------------
    # FASTAPI
    # --------------------------------------------------------

    response = requests.post(
        API_URL,
        json=raw_product,
        timeout=30
    )

    if response.status_code != 200:

        print(
            f"API ERROR at product {index}: "
            f"HTTP {response.status_code}"
        )

        print(response.text)

        raise RuntimeError(
            f"FastAPI failed at product index {index}"
        )

    api_result = response.json()

    # --------------------------------------------------------
    # COMPARE
    # --------------------------------------------------------

    risk_match = (
        direct_result["risk_score"]
        == api_result["risk_score"]
    )

    level_match = (
        direct_result["risk_level"]
        == api_result["risk_level"]
    )

    anomaly_score_match = (
        direct_result["anomaly_score"]
        == api_result["anomaly_score"]
    )

    indicator_score_match = (
        direct_result["indicator_score"]
        == api_result["indicator_score"]
    )

    anomaly_flag_match = (
        direct_result["is_anomaly"]
        == api_result["is_anomaly"]
    )

    all_match = (
        risk_match
        and level_match
        and anomaly_score_match
        and indicator_score_match
        and anomaly_flag_match
    )

    parity_results.append({

        "index":
            index,

        "parent_asin":
            raw_product.get(
                "parent_asin"
            ),

        "risk_score_direct":
            direct_result["risk_score"],

        "risk_score_api":
            api_result["risk_score"],

        "risk_level_direct":
            direct_result["risk_level"],

        "risk_level_api":
            api_result["risk_level"],

        "anomaly_score_direct":
            direct_result["anomaly_score"],

        "anomaly_score_api":
            api_result["anomaly_score"],

        "indicator_score_direct":
            direct_result["indicator_score"],

        "indicator_score_api":
            api_result["indicator_score"],

        "is_anomaly_direct":
            direct_result["is_anomaly"],

        "is_anomaly_api":
            api_result["is_anomaly"],

        "parity":
            all_match
    })


parity_df = pd.DataFrame(
    parity_results
)


# ============================================================
# FINAL PARITY REPORT
# ============================================================

print("=" * 80)
print("PHASE 5B — MULTI-PRODUCT FASTAPI PARITY VALIDATION")
print("=" * 80)

print(
    f"\nProducts tested: {len(parity_df)}"
)

print(
    f"Exact parity matches: "
    f"{parity_df['parity'].sum()}"
)

print(
    f"Parity failures: "
    f"{(~parity_df['parity']).sum()}"
)

parity_percentage = (
    parity_df["parity"].mean()
    * 100
)

print(
    f"Parity percentage: "
    f"{parity_percentage:.2f}%"
)


# ------------------------------------------------------------
# Display mismatches if any
# ------------------------------------------------------------

mismatches = parity_df[
    ~parity_df["parity"]
]

if len(mismatches) == 0:

    print("\n✓ PERFECT PARITY")
    print(
        "Direct Python engine and FastAPI "
        "returned identical results for "
        "every tested product."
    )

else:

    print("\n⚠ PARITY MISMATCHES FOUND")

    display(
        mismatches
    )


# ------------------------------------------------------------
# Show sample results
# ------------------------------------------------------------

print("\nSAMPLE RESULTS")
print("-" * 80)

display(
    parity_df[
        [
            "parent_asin",
            "risk_score_direct",
            "risk_score_api",
            "risk_level_direct",
            "risk_level_api",
            "parity"
        ]
    ].head(10)
)

print("\n" + "=" * 80)
print("MULTI-PRODUCT PARITY TEST COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# PHASE 5B — FASTAPI CONTRACT & EDGE-CASE AUDIT
# ============================================================

import requests
import json
import pandas as pd
from IPython.display import display


API_BASE = "http://127.0.0.1:8000"


# ============================================================
# TEST 1 — HEALTH ENDPOINT
# ============================================================

print("=" * 80)
print("TEST 1 — HEALTH ENDPOINT")
print("=" * 80)

response = requests.get(
    f"{API_BASE}/health",
    timeout=10
)

print("HTTP Status:", response.status_code)
print("Response:", response.json())

assert response.status_code == 200
assert response.json()["status"] == "healthy"

print("✓ HEALTH CHECK PASSED")


# ============================================================
# TEST 2 — ROOT ENDPOINT
# ============================================================

print("\n" + "=" * 80)
print("TEST 2 — ROOT ENDPOINT")
print("=" * 80)

response = requests.get(
    f"{API_BASE}/",
    timeout=10
)

print("HTTP Status:", response.status_code)
print("Response:", response.json())

assert response.status_code == 200

print("✓ ROOT ENDPOINT PASSED")


# ============================================================
# TEST 3 — VALID PRODUCT
# ============================================================

print("\n" + "=" * 80)
print("TEST 3 — VALID PRODUCT")
print("=" * 80)

valid_product = {
    "parent_asin": "TEST001",

    "title":
        "Wireless Bluetooth Headphones",

    "description":
        "High quality wireless headphones "
        "with long battery life.",

    "features": [
        "Bluetooth",
        "Noise cancellation",
        "Long battery life"
    ],

    "categories": [
        "Electronics",
        "Headphones"
    ],

    "images": {
        "large": [
            "https://example.com/image.jpg"
        ]
    },

    "videos": {
        "url": []
    },

    "seller":
        "Test Electronics",

    "price":
        "49.99",

    "average_rating":
        4.2,

    "rating_number":
        150
}

response = requests.post(
    f"{API_BASE}/predict",
    json=valid_product,
    timeout=30
)

print("HTTP Status:", response.status_code)

assert response.status_code == 200

valid_result = response.json()

print(
    json.dumps(
        valid_result,
        indent=4
    )
)

assert 0 <= valid_result["risk_score"] <= 100
assert 0 <= valid_result["anomaly_score"] <= 100
assert 0 <= valid_result["indicator_score"] <= 100

assert valid_result["risk_level"] in [
    "LOW",
    "MEDIUM",
    "HIGH",
    "CRITICAL"
]

print("✓ VALID PRODUCT TEST PASSED")


# ============================================================
# TEST 4 — MINIMAL / MISSING METADATA
# ============================================================

print("\n" + "=" * 80)
print("TEST 4 — MISSING METADATA")
print("=" * 80)

minimal_product = {}

response = requests.post(
    f"{API_BASE}/predict",
    json=minimal_product,
    timeout=30
)

print("HTTP Status:", response.status_code)

assert response.status_code == 200

minimal_result = response.json()

print(
    json.dumps(
        minimal_result,
        indent=4
    )
)

assert 0 <= minimal_result["risk_score"] <= 100

print("✓ MISSING METADATA TEST PASSED")


# ============================================================
# TEST 5 — SUSPICIOUS PRODUCT
# ============================================================

print("\n" + "=" * 80)
print("TEST 5 — SUSPICIOUS PRODUCT")
print("=" * 80)

suspicious_product = {

    "title":
        "Amazing Product",

    "description":
        "Best product ever.",

    "features": [],

    "categories": [
        "Electronics"
    ],

    "images": {},

    "videos": {},

    "seller":
        "",

    "price":
        "9.99",

    "average_rating":
        5.0,

    "rating_number":
        2
}

response = requests.post(
    f"{API_BASE}/predict",
    json=suspicious_product,
    timeout=30
)

print("HTTP Status:", response.status_code)

assert response.status_code == 200

suspicious_result = response.json()

print(
    json.dumps(
        suspicious_result,
        indent=4
    )
)

assert suspicious_result["risk_level"] in [
    "HIGH",
    "CRITICAL"
]

print("✓ SUSPICIOUS PRODUCT TEST PASSED")


# ============================================================
# TEST 6 — INVALID FIELD TYPE
# ============================================================

print("\n" + "=" * 80)
print("TEST 6 — INVALID FIELD TYPE")
print("=" * 80)

invalid_product = {

    "title":
        "Test Product",

    "average_rating":
        "NOT_A_NUMBER",

    "rating_number":
        10
}

response = requests.post(
    f"{API_BASE}/predict",
    json=invalid_product,
    timeout=30
)

print("HTTP Status:", response.status_code)

print(
    response.text
)

assert response.status_code == 422

print("✓ INVALID INPUT TEST PASSED")


# ============================================================
# TEST 7 — RESPONSE SCHEMA
# ============================================================

print("\n" + "=" * 80)
print("TEST 7 — RESPONSE SCHEMA")
print("=" * 80)

required_fields = [
    "risk_score",
    "risk_level",
    "is_anomaly",
    "anomaly_score",
    "indicator_score",
    "isolation_forest",
    "risk_weights",
    "explanations"
]

missing_fields = [
    field
    for field in required_fields
    if field not in valid_result
]

print(
    "Missing fields:",
    missing_fields
)

assert len(missing_fields) == 0

print("✓ RESPONSE SCHEMA PASSED")


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("PHASE 5B — API CONTRACT & EDGE-CASE AUDIT COMPLETE")
print("=" * 80)

print("""
✓ Health endpoint
✓ Root endpoint
✓ Valid product
✓ Missing metadata
✓ Suspicious product
✓ Invalid input handling
✓ Response schema

PHASE 5B API CONTRACT STATUS: PASS
""")
```
